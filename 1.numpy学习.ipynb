# 一、ndarray
##  1.ndarray的特性
### 多维性


```python
import numpy as np
```

```python
import pandas as  pd
```

```python
arr = np.array(5)  # 创建0维的ndarray数组
print(arr)
print('arr的维度：', arr.ndim) # number of dimensions
```

```python
arr = np.array([1,2,3])  # 创建1维的ndarray数组
print(arr)
print('arr的维度：', arr.ndim) # number of dimensions
```

```python
arr = np.array([[1,2,3],[4,5,6]])  # 创建2维的ndarray数组
print(arr)
print('arr的维度：', arr.ndim) # number of dimensions
```

### 同质性

```python
arr = np.array([1,'Hello']) # 不同的数据类型会被强制转换成相同的数据类型
print(arr)
```

```python
arr = np.array([1,2.5]) # 不同的数据类型会被强制转换成相同的数据类型
print(arr)
```

## 2.ndarray的属性 

```python
arr = np.array(1)
print(arr)
print('数组的形状：', arr.shape)
print('数组的维度：', arr.ndim)
print('元素的个数：', arr.size)
print('元素的数据类型：', arr.dtype)
```

```python
arr = np.array([1,2.5,3])
print(arr)
print('数组的形状：', arr.shape)
print('数组的维度：', arr.ndim)
print('元素的个数：', arr.size)
print('元素的数据类型：', arr.dtype)
print('元素的转置：',  arr.T)
```

```python
arr = np.array([[1,2,3],[4,5,6]])
print(arr)
print('数组的形状：', arr.shape)
print('数组的维度：', arr.ndim)
print('元素的个数：', arr.size)
print('元素的数据类型：', arr.dtype)
print('元素的转置：',  arr.T)
```

## 3.ndarray的创建

```python
# 基础的创建方法
list1  = [4,5,6]
arr = np.array(list1,dtype=np.float64)
print(arr.ndim)  # 属性
print(arr)
```

```python
#  copy
arr1 = np.copy(arr)  # 元素跟原始的数组相同，但是不是一个数组了
print(arr1)
arr1[0] = 8
print(arr1)
print(arr)
```

```python
# 预定义形状
# 全0  全1 未初始化 固定值

# 全0
arr  = np.zeros((2,3),dtype  = int)
print(arr)
print(arr.dtype)
```

```python
# 预定义形状
# 全0  全1 未初始化 固定值

# 全0
arr  = np.zeros((200,),dtype  = int)
print(arr)
print(arr.ndim)  # 这是一维的
```

```python
#  全1
arr =  np.ones((5,8),dtype =  int)
print(arr)
```

```python
# 未初始化
arr = np.empty((2,3))
print(arr)
```

```python
# 第二次改一下格式 初始化后的
arr = np.empty((4,3))
print(arr)
```

```python
arr = np.full((3,4),2025)
print(arr)
```

```python
arr1 = np.zeros_like(arr)
print(arr1)
arr1 = np.empty_like(arr)
print(arr1)
arr1 = np.ones_like(arr)
print(arr1)
arr1 = np.full_like(arr,2026)
print(arr1)
```

```python
# 等差数列 2 4 6 8 10
arr =  np.arange(2,11,2) # start,end,step（步长）
print(arr)
```

```python
#等间隔数列
arr = np.linspace(0,100,5,dtype=int) #  第三个是取几份
print(arr)
arr =  np.arange(0,101,25)
print(arr)
```

```python
# 对数间隔数列  （其实是指数 等比数列）不写默认是10
arr =  np.logspace(0,4,3,base=2)
print(arr)
```

```python
# 特殊矩阵
# 单位矩阵：主对角线上的数字为1，其他的数字为0
arr = np.eye(3,4,dtype =int)
print(arr)
```

```python
# 对角矩阵：主对角线上非零的0，其他的数字为0
arr = np.diag([5,1,2,3])
print(arr)
```

```python
# 随机数组的生成
# 生成0到1之间的随机浮点数（均匀分布）
arr = np.random.rand(2,3)
print(arr)
```

```python
# 生成指定范围区间的随机浮点数 （均匀分布）
arr =  np.random.uniform(3,6,(2,3))
print(arr)
```

```python
#  生成指定范围区间的随机整数
arr = np.random.randint(3,30,(2,3))
print(arr)
  
```

```python
# 生成随机数列（正态分布）（-3到3之间）
# 两边概率小 中间大 
arr =  np.random.randn(2,3)
print(arr)
```

```python
# 设置随机种子
np.random.seed(20)
arr = np.random.randint(1,10,(2,5))
print(arr)
```
